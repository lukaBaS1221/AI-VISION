```python
PEOPLE = ['man', 'men', 'woman', 'women', 'person', 'people', 'boy', 'boys', 'girl', 'girls', 'child', 'children']
OUTDOOR = ["forest", "park", "street", "bench", "road", "highway", "alley", "sidewalk", "city", "town", "village", "neighborhood", "downtown", "urban area", "suburban area", "rural area", "countryside", "field", "meadow", "grassland", "farmland", "farm", "orchard", "vineyard", "mountain", "hill", "valley", "cliff", "plateau", "canyon", "desert", "dune", "oasis", "beach", "shore", "coast", "seaside", "bay", "harbor", "port", "island", "lake", "river", "stream", "waterfall", "pond", "wetland", "marsh", "swamp", "jungle", "rainforest", "savanna", "trail", "path", "hiking trail", "campground", "campsite", "picnic area", "playground", "garden", "botanical garden", "zoo", "stadium", "sports field", "soccer field", "basketball court", "tennis court", "skate park", "plaza", "square", "market", "outdoor market", "festival area", "construction site", "parking lot", "gas station", "bridge", "tunnel entrance", "railway", "train station exterior", "bus stop", "airport exterior", "runway", "industrial area", "factory exterior", "warehouse exterior", "dock", "pier", "boardwalk", "resort", "outdoor cafe", "restaurant terrace", "courtyard", "campus", "schoolyard", "playfield"]

INDOOR = ["living room", "bedroom", "kitchen", "bathroom", "dining room", "hallway", "corridor", "staircase", "basement", "attic", "garage", "laundry room", "home office", "study room", "closet", "pantry", "apartment", "house interior", "hotel room", "hotel lobby", "office", "conference room", "meeting room", "classroom", "lecture hall", "library", "computer lab", "cafeteria", "restaurant interior", "cafe interior", "bar interior", "shopping mall", "store", "supermarket", "hospital room", "clinic", "waiting room", "gym", "movie theater", "museum", "art gallery", "airport terminal", "train station interior", "elevator", "warehouse interior", "factory interior", "server room"]

ANIMALS = ["dog", "dogs", "cat", "cats", "bird", "birds", "horse", "horses", "cow", "cows", "sheep", "goat", "goats", "pig", "pigs", "rabbit", "rabbits", "deer", "fox", "foxes", "wolf", "wolves", "bear", "bears", "lion", "lions", "tiger", "tigers", "elephant", "elephants", "monkey", "monkeys", "fish", "dolphin", "dolphins", "shark", "sharks", "whale", "whales", "snake", "snakes", "lizard", "lizards", "turtle", "turtles", "frog", "frogs"]

ACTIONS = ["walking", "running", "standing", "sitting", "lying down", "jumping", "climbing", "talking", "eating", "drinking", "working", "studying", "reading", "writing", "typing", "using phone", "playing", "exercising", "driving", "riding", "cooking", "cleaning", "shopping", "waiting"]

PEOPLE_PATTERN = re.compile(r'(\d+)\s*(man|men|woman|women|person|people|boy|boys|girl|girls|child|children)')
ANIMAL_PATTERN = re.compile(r'(\d+)\s*(dog|dogs|cat|cats|bird|birds|horse|horses|cow|cows|sheep|goat|goats|pig|pigs|rabbit|rabbits|deer|fox|foxes|wolf|wolves|bear|bears|lion|lions|tiger|tigers|elephant|elephants|monkey|monkeys|fish|dolphin|dolphins|shark|sharks|whale|whales|snake|snakes|lizard|lizards|turtle|turtles|frog|frogs)')

def analyze_image(image):

    prompt = (
        "Describe this image concisely in 2 to 3 sentences with a maximum of 30 words . "
        "Use numeric counts (e.g., 3, not three,)(if there is one subject dont mention the value, otherwise use numeric counts). "
        "Use general people terms. "
        "Format exactly as: "
        "Description: ... People: X. Animals: Y. Scene: indoor/outdoor. Activity: ..."
    )

    response = gemini_model.generate_content([image, prompt])
    text = response.text.lower()

    basic = ""
    people_count = 0
    animal_count = 0
    scene = "unknown"
    activity = "unknown"

    desc_match = re.search(r'description:\s*(.*?)(?:people:|animals:|scene:|activity:|$)', text, re.DOTALL)
    if desc_match:
        basic = desc_match.group(1).strip()
    else:
        basic = text.strip()

    people_match = re.search(r'people:\s*(\d+)', text)
    if people_match:
        people_count = int(people_match.group(1))
    else:
        matches = PEOPLE_PATTERN.findall(basic)
        people_count = sum(int(m[0]) for m in matches) if matches else 0

    animal_match = re.search(r'animals:\s*(\d+)', text)
    if animal_match:
        animal_count = int(animal_match.group(1))
    else:
        matches = ANIMAL_PATTERN.findall(basic)
        animal_count = sum(int(m[0]) for m in matches) if matches else 0

    scene_match = re.search(r'scene:\s*(indoor|outdoor)', text)
    if scene_match:
        scene = scene_match.group(1)
    else:
        outdoor_hits = sum(w in basic for w in OUTDOOR)
        indoor_hits = sum(w in basic for w in INDOOR)
        if outdoor_hits > indoor_hits:
            scene = "outdoor"
        elif indoor_hits > outdoor_hits:
            scene = "indoor"

    activity_match = re.search(r'activity:\s*(.*)', text)
    if activity_match:
        activity = activity_match.group(1).strip()
    else:
        activity = next((a for a in ACTIONS if a in basic), "static")

    enhanced = (
        f"This is an {scene} scene. "
        f"It features {people_count} person(s) and {animal_count} animal(s). "
        f"Main activity: {activity}. "
        f"Description: {basic.capitalize()}."
    )

    return {
        "basic": basic,
        "enhanced": enhanced,
        "people": people_count,
        "animals": animal_count,
        "scene": scene,
        "activity": activity
    }
```

```python
test_idx = 73
result = analyze_image(dataset[test_idx]['image'])


plt.imshow(dataset[test_idx]['image'])
plt.axis('off')
plt.title(f"Image {test_idx} - Analysis")
plt.show()

print(f"Basic Description: {result['basic']}")
print(f"\nEnhanced Description: {result['enhanced']}")
print(f"\nAnalysis: {result['people']} people, {result['animals']} animals, {result['scene']} scene, activity: {result['activity']}")
```

```python
from deep_translator import GoogleTranslator

def translate_to_georgian(text):

    translator = GoogleTranslator(source='en', target='ka')
    return translator.translate(text)


result = analyze_image(dataset[test_idx]['image'])

georgian_enhanced = translate_to_georgian(result['enhanced'])

print(f"English Enhanced: {result['enhanced']}")
print(f"\nქართული თარგმანი: {georgian_enhanced}")
```

```python
import io
import gradio as gr
from google.cloud import texttospeech



def english_tts(text):
    """
    Generate English audio using Google Cloud TTS.
    Returns path to saved .wav file.
    """
    client = texttospeech.TextToSpeechClient()
    input_text = texttospeech.SynthesisInput(text=text)
    voice = texttospeech.VoiceSelectionParams(
        language_code="en-US",
        name="en-US-Standard-A"
    )
    audio_config = texttospeech.AudioConfig(
        audio_encoding=texttospeech.AudioEncoding.LINEAR16
    )
    response = client.synthesize_speech(
        input=input_text,
        voice=voice,
        audio_config=audio_config
    )

    audio_file = "english_audio.wav"
    with open(audio_file, "wb") as f:
        f.write(response.audio_content)
    return audio_file


def gradio_pipeline(image):

    result = analyze_image(image)
    english_text = result["enhanced"]

    georgian_text = translate_to_georgian(english_text)

    audio_file = english_tts(english_text)

    return english_text, georgian_text, audio_file


with gr.Blocks(title="Image → English Audio") as demo:
    gr.Markdown("## 🖼️ Image → English → Georgian → 🔊 Audio")



    image_input = gr.Image(type="pil", label="Upload Image")

    english_out = gr.Textbox(label="English Description", lines=4)
    georgian_out = gr.Textbox(label="ქართული აღწერა", lines=4)
    audio_out = gr.Audio(label="English Audio", type="filepath")

    run_btn = gr.Button("Analyze Image")
    run_btn.click(
        fn=gradio_pipeline,
        inputs=image_input,
        outputs=[english_out, georgian_out, audio_out]

    )

    gr.Image("TBC.png", label="Logo", interactive=False)

demo.launch(debug=True)
```
